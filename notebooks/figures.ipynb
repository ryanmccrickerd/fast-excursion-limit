# Generate all article figures

Drives the 5 base figure notebooks via [papermill](https://papermill.readthedocs.io/). `figure-5.ipynb` is run in three ways for figures 5-7.

`num_paths` is a papermill parameter, overridden in the `figures.yml` workflow.

```python
import time
from pathlib import Path

import papermill as pm

FIGURES_DIR = Path("figures")
SCRATCH_DIR = Path("../scratch")
SCRATCH_DIR.mkdir(parents=True, exist_ok=True)


def run(notebook, parameters=None):
    default_output = notebook.removesuffix(".ipynb") + ".pdf"
    label = (parameters or {}).get("filename", default_output)
    print(f"--- running {notebook} -> {label} ---", flush=True)
    start = time.perf_counter()
    pm.execute_notebook(
        FIGURES_DIR / notebook,
        SCRATCH_DIR / f"{notebook.removesuffix('.ipynb')}-executed.ipynb",
        parameters=parameters or {},
    )
    print(f"--- done: {label} ({time.perf_counter() - start:.0f}s) ---", flush=True)
```

## Figures 1-4 (run unmodified)

```python
for n in [1, 2, 3, 4]:
    run(f"figure-{n}.ipynb")
```
