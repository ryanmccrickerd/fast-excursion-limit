# Figure 3 — FEH paths at the correlation boundaries

```python
import matplotlib.pyplot as plt
import numpy as np

from fast_excursion_limit import plot_config
from fast_excursion_limit.fast_excursion_heston import FastExcursionHeston

plot_config.set_style()

SAVE_PLOT = True
```

## Parameters

```python
from fast_excursion_limit import defaults

rhos = [1, -1]

num_paths = 2

seed = defaults.SEED
maturity = defaults.MATURITY_PATHWISE
step_size = defaults.DELTA_SCALE_PATHWISE * maturity
```

## Simulate paths

```python
model = FastExcursionHeston.example()

paths = {}
for rho in rhos:
    np.random.seed(seed)
    model.rho = rho
    paths[rho] = [
        model.simulate(maturity=maturity, step_size=step_size) for _ in range(num_paths)
    ]
```

## Plot

```python
excursion_labels = {1: "upwards excursions only", -1: "downwards excursions only"}

fig, axes = plt.subplots(
    1, len(rhos), figsize=plot_config.TWO_PANEL_FIGSIZE, sharey=True
)
for ax, rho in zip(axes, rhos):
    ax.set_box_aspect(1)
    for Z, Y in paths[rho]:
        ax.plot(Y, Z)
    ax.set_title(rf"$\rho = {rho:+d}$ ({excursion_labels[rho]})")
    ax.set_xlabel("Time")
axes[0].set_ylabel("Price")
fig.tight_layout()
if SAVE_PLOT:
    fig.savefig(plot_config.PLOTS_DIR / "figure-3.pdf")
```
