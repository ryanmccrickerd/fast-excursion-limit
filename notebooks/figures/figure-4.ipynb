# Figure 4 — FEH paths at the vol-of-vol boundaries

```python
import matplotlib.pyplot as plt
import numpy as np

from fast_excursion_limit import plot_config
from fast_excursion_limit.fast_excursion_heston import FastExcursionHeston

plot_config.set_style()

SAVE_PLOT = True
```

## Parameters

```python
from fast_excursion_limit import defaults

gammas = [0, 2]

num_paths = 2

seed = defaults.SEED
maturity = defaults.MATURITY_PATHWISE
step_size = defaults.DELTA_SCALE_PATHWISE * maturity
```

## Simulate paths

```python
model = FastExcursionHeston.example()

paths = {}
for gamma in gammas:
    np.random.seed(seed)
    model.gamma = gamma
    paths[gamma] = [
        model.simulate(maturity=maturity, step_size=step_size) for _ in range(num_paths)
    ]
```

## Plot

```python
gamma_labels = {0: "Black-Scholes, no excursions", 2: "extreme case"}

fig, axes = plt.subplots(
    1, len(gammas), figsize=plot_config.TWO_PANEL_FIGSIZE, sharey=True
)
for ax, gamma in zip(axes, gammas):
    ax.set_box_aspect(1)
    for Z, Y in paths[gamma]:
        ax.plot(Y, Z)
    ax.set_title(rf"$\gamma = {gamma}$ ({gamma_labels[gamma]})")
    ax.set_xlabel("Time")
axes[0].set_ylabel("Price")
fig.tight_layout()
if SAVE_PLOT:
    fig.savefig(plot_config.PLOTS_DIR / "figure-4.pdf")
```
